```python
def is_prime(n):
    """Check if a number is prime."""
    if n < 2:
        return False
    for i in range(2, int(n ** 0.5) + 1):
        if n % i == 0:
            return False
    return True
```

```python
print(is_prime(7))
```

```python
import unittest

class TestIsPrime(unittest.TestCase):
    def test_negative_number(self):
        self.assertEqual(is_prime(-1), False)
        self.assertNotEqual(is_prime(-1), True)
        self.assertTrue(not is_prime(-1))

    def test_edge_case_zero(self):
        self.assertEqual(is_prime(0), False)
        self.assertNotEqual(is_prime(0), True)
        self.assertTrue(not is_prime(0))

    def test_edge_case_one(self):
        self.assertEqual(is_prime(1), False)
        self.assertNotEqual(is_prime(1), True)
        self.assertTrue(not is_prime(1))

    def test_small_prime(self):
        self.assertEqual(is_prime(2), True)
        self.assertNotEqual(is_prime(2), False)
        self.assertTrue(is_prime(2))

    def test_small_non_prime(self):
        self.assertEqual(is_prime(4), False)
        self.assertNotEqual(is_prime(4), True)
        self.assertTrue(not is_prime(4))

    def test_medium_prime(self):
        self.assertEqual(is_prime(29), True)
        self.assertNotEqual(is_prime(29), False)
        self.assertTrue(is_prime(29))

    def test_medium_non_prime(self):
        self.assertEqual(is_prime(30), False)
        self.assertNotEqual(is_prime(30), True)
        self.assertTrue(not is_prime(30))

    def test_large_prime(self):
        self.assertEqual(is_prime(7919), True)
        self.assertNotEqual(is_prime(7919), False)
        self.assertTrue(is_prime(7919))

    def test_large_non_prime(self):
        self.assertEqual(is_prime(8000), False)
        self.assertNotEqual(is_prime(8000), True)
        self.assertTrue(not is_prime(8000))
```

```python
if __name__ == "__main__":
    unittest.main(argv=['first-arg-is-ignored'], exit=False)
```
